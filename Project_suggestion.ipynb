**Diabetic Retinopathy (DR)** is the leading cause of blindness among working-age individuals in developed nations, affecting over 93 million people globally. This eye disease is a complication of long-standing diabetes, which is itself a growing epidemic. According to the CDC, 29.1 million Americans have diabetes, and the WHO estimates the global prevalence at 347 million. Of these, approximately 40–45% of diabetic Americans show some stage of DR. Without timely detection and intervention, DR can progress to severe vision impairment or blindness, often with few symptoms until it is too late for effective treatment.

Detection of DR currently relies on manual examination of digital color fundus photographs of the retina by trained clinicians. This approach, while effective, is plagued by several challenges:

- **Time-Intensive Process**: Reviews often take days, delaying communication and treatment.
- **Resource Constraints**: Detection requires specialized expertise and equipment, which are often unavailable in under-resourced areas where diabetes prevalence is high.
- **Growing Burden**: The increasing number of people with diabetes is outpacing the infrastructure needed to prevent blindness caused by DR.

To address these challenges, there is a critical need for a scalable, automated system for DR detection. Such a solution could significantly improve outcomes by enabling early diagnosis and timely treatment.

```python
def plot_class_distribution(df):
    """
    Plots a 'Pie of a Pie' chart showing the distribution of 'level' classes,
    with a simplified main chart of healthy vs sick, and a breakdown of 'sick' cases (levels 1 to 4).

    Parameters:
        df (pd.DataFrame): DataFrame containing at least the 'level' column.
    """
    # Calculate healthy and sick counts
    healthy_count = df[df['level'] == 0].shape[0]
    sick_count = df[df['level'] > 0].shape[0]

    # Main Pie Chart Data: Healthy vs Sick
    sizes_main = [healthy_count, sick_count]
    labels_main = ['Healthy (0)', 'Sick (1-4)']
    colors_main = ['#6baed6', '#fd8d3c']  # Blue for Healthy, Orange for Sick
    explode_main = [0, 0.1]  # Slightly separate the "Sick" category

    # Sick Cases Pie Chart Data (Levels 1 to 4)
    sick_levels = [1, 2, 3, 4]
    sick_labels = ['Mild (1)', 'Moderate (2)', 'Severe (3)', 'Proliferative (4)']
    sick_sizes = [df[df['level'] == level].shape[0] for level in sick_levels]
    colors_sick = plt.cm.Paired(np.arange(1, len(sick_levels)+1))

    # Create figure and subplots
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 7))

    # Main Pie Chart: Healthy vs Sick
    ax1.pie(sizes_main,
            labels=labels_main,
            autopct='%1.1f%%',
            startangle=90,
            colors=colors_main,
            explode=explode_main,
            wedgeprops={'linewidth': 1, 'edgecolor': 'white'})
    ax1.set_title("Class Distribution: Healthy vs Sick")
    ax1.axis('equal')  # Ensures pie is circular

    # 'Sick' Cases Pie Chart: Levels 1 to 4
    ax2.pie(sick_sizes,
            labels=sick_labels,
            autopct='%1.1f%%',
            colors=colors_sick,
            wedgeprops={'linewidth': 1, 'edgecolor': 'white'})
    ax2.set_title("Breakdown of Sick Cases (Levels 1-4)")
    ax2.axis('equal')  # Ensures pie is circular
    plt.tight_layout()
    plt.show()

plot_class_distribution(df)

pd.DataFrame.from_dict(dict(Counter(df['level'])), orient='index', columns=['Count']).reindex([0, 1, 2, 3, 4]).rename(index={0: 'no DR', 1: 'Mild', 2: 'Moderate', 3: 'Severe', 4: 'Proliferative'}).rename_axis("Severity Level")
```

```python
def show_examples_per_class(df, images_folder, num_examples=5):
    """
    Displays a few sample images from each class.

    Parameters:
        df (pd.DataFrame): Must have columns 'image' and 'level'
        images_folder (str): Path to where images are stored (each image named {image_id}.jpeg)
        num_examples (int): Number of examples to display per class
    """
    # Unique classes sorted for consistent ordering
    classes = sorted(df['level'].unique())

    # Figure setup: one row per class, `num_examples` columns
    fig, axes = plt.subplots(nrows=len(classes), ncols=num_examples, figsize=(4 * num_examples, 4 * len(classes)))
    if len(classes) == 1:
        axes = [axes]  # Ensure we can iterate if there's only one class

    for row_idx, cls in enumerate(classes):
        # Filter the dataframe for the current class
        df_cls = df[df['level'] == cls]

        # Randomly sample a few images (up to num_examples)
        sample_df = df_cls.sample(n=min(num_examples, len(df_cls)), random_state=random_state)

        for col_idx, (_, row) in enumerate(sample_df.iterrows()):
            image_id = row['image']
            img_path = os.path.join(images_folder, f"{image_id}.jpeg")

            # Read and convert to RGB for plotting
            bgr_img = cv2.imread(img_path)
            if bgr_img is not None:
                rgb_img = cv2.cvtColor(bgr_img, cv2.COLOR_BGR2RGB)
            else:
                # If image is not found or can't be read, create a placeholder
                rgb_img = None

            ax = axes[row_idx][col_idx] if len(classes) > 1 else axes[col_idx]
            ax.axis('off')

            if rgb_img is not None:
                ax.imshow(rgb_img)
                ax.set_title(f"Class {cls} (ID: {image_id})")
            else:
                ax.text(0.5, 0.5, "Image Not Found", ha='center', va='center')
                ax.set_title(f"Class {cls}\nMissing: {image_id}")

    plt.tight_layout()
    plt.show()

show_examples_per_class(df, images_path)
```

```python
def plot_image_sizes(df, images_folder, max_workers=os.cpu_count() if os.cpu_count() else 4):
    """
    Plots a scatter chart with:
        - x-axis = image width
        - y-axis = image height
        - hue = class (level)
        - size = frequency of (width, height, class)

    Additionally, adds a new 'size' column to the DataFrame containing (width, height) tuples.

    Args:
        df (pd.DataFrame): DataFrame containing 'image' and 'level' columns.
        images_folder (str): Path to folder containing images.
        max_workers (int, optional): Maximum number of threads to use. Defaults to CPU count or 4.
    """

    def process_image(args):
        """
        Process a single image and return its dimensions along with image_id and level.

        Args:
            args (tuple): (image_id, level, images_folder)

        Returns:
            tuple or None: (image_id, width, height, level) or None if failed
        """
        image_id, level, images_folder = args
        img_path = os.path.join(images_folder, f"{image_id}.jpeg")

        try:
            # Read the image using OpenCV
            bgr_img = cv2.imread(img_path)
            if bgr_img is not None:
                height, width, _ = bgr_img.shape
                return (image_id, width, height, level)
            else:
                print(f"Warning: Image not found or unable to read: {img_path}")
        except Exception as e:
            print(f"Error processing image {image_id}: {e}")
        return None

    # Prepare arguments for parallel processing
    args_list = [(row['image'], row['level'], images_folder) for _, row in df.iterrows()]

    # Process images using ThreadPoolExecutor
    sizes = []
    with ThreadPoolExecutor(max_workers=max_workers) as executor:
        futures = [executor.submit(process_image, args) for args in args_list]

        # Show progress bar while processing
        for future in tqdm(as_completed(futures), total=len(futures), desc="Processing images", unit="images"):
            result = future.result()
            if result is not None:
                sizes.append(result)

    # Convert to DataFrame
    sizes_df = pd.DataFrame(sizes, columns=['image', 'width', 'height', 'level'])

    # Create a 'size' column as tuple (width, height)
    sizes_df['size'] = list(zip(sizes_df['width'], sizes_df['height']))

    # Create a mapping from (image, level) to size tuple
    size_mapping = sizes_df.set_index(['image', 'level'])['size'].to_dict()

    # Add 'size' column to the original df in place
    df['size'] = df.apply(lambda row: size_mapping.get((row['image'], row['level'])), axis=1)

    # Now, for plotting, count frequency of each (width, height, level) combination
    # This is used to determine the size of each scatter point
    sizes_agg = sizes_df.groupby(['width', 'height', 'level']).size().reset_index(name='count')


    plt.figure(figsize=(12, 8))
    scatter_plot = sns.scatterplot(data=sizes_agg, x='width', y='height',
                                   hue='level', size='count', sizes=(100, 1000),
                                   palette='viridis', alpha=0.7, edgecolor='w', linewidth=0.5)
    scatter_plot.set_title("Image Dimensions by Class", fontsize=16)
    scatter_plot.set_xlabel("Width (pixels)", fontsize=14)
    scatter_plot.set_ylabel("Height (pixels)", fontsize=14)

    # Enhance legend placement and appearance
    handles, labels = scatter_plot.get_legend_handles_labels()
    # Exclude the first handle if it's the scatter points itself
    plt.legend(handles=handles[1:], labels=labels[1:], bbox_to_anchor=(1.05, 1), loc='upper left', title="Level & Count")
    plt.grid(True, alpha=0.7)
    plt.tight_layout()
    plt.show()

    plt.figure(figsize=(12, 6))
    sns.boxplot(data=sizes_df[['width', 'height']], orient="h", palette="coolwarm")
    plt.title("Box Plot of Image Dimensions")
    plt.xlabel("Pixels")
    plt.show()


plot_image_sizes(df, images_path)
```

The following script analyzes retinal images and extracts geometric features such as the crop centers, bounding box, and circularity to provide insights into the shape and structure of the retina. Each retina image is converted to grayscale and thresholded to create a binary mask that isolates the retina from the background. Using contour detection, the largest contour is identified, assuming it corresponds to the retina. From this, the center of the retina is calculated and normalized relative to the image dimensions.

To better approximate the retina's shape, the script uses the *convex hull*. A convex hull is the smallest convex boundary that can encompass the contour, essentially "smoothing out" irregularities or noise in the shape. This is crucial because the raw contours often contain small indentations or fluctuations that can distort metrics like circularity or area ratios. By using the convex hull, the script provides a more robust and accurate representation of the retina's geometry, specifically addressing the common protrusion at the top corner of the retinal images. The use of the convex hull ensures that these distortions do not negatively impact the calculations, making it a key component of this analysis. An example of a convex hull can be seen below, where the blue outline represents the convex hull. It can be imagined as releasing a rubber band around a group of fixed points.

![Convex Hull example](ConvexHull.svg)

The script then calculates several key metrics:
- **bounding box:** The bounding box is the smallest rectangle that completely contains the retina's largest contour. This is later used to crop the images efficiently, ensuring focus on the retina itself.
- **Circularity:** Using the convex hull, circularity is refined to reflect how closely the retina shape resembles a perfect circle. The formula for circularity is:
  $$
  \text{Circularity} = \frac{4 \cdot \pi \cdot \text{Hull Area}}{\text{Hull Perimeter}^2}
  $$
  Circularity compares the area of the convex hull to its perimeter. The factor $4\pi$ accounts for the geometry of a perfect circle, where $Area = \pi r^2$ and $Perimeter=2\pi r$. For a circle, this formula simplifies to 1 because all other shapes, having a larger perimeter relative to their area, will have a lower value. Thus, the circularity metric is particularly sensitive to irregularities, and the convex hull ensures more reliable results by smoothing out noise or distortions in the contour.
- **center_y & center_x:** These represent the normalized coordinates of the retina's center, calculated using the moments of the largest contour. Normalization ensures that the center coordinates are expressed as fractions of the image's width and height, making them independent of the image's resolution.


To make the results more intuitive, the script visualizes the data in a scatter plot. This plot highlights the normalized crop centers of the retina, showing their spatial distribution across the image space. The circularity of each retina is represented using color intensity, providing an additional layer of information.

```python
def analyze_retina_images(images_folder, df, max_workers=os.cpu_count() if os.cpu_count() else 4):
    """
    Analyzes retina images to calculate bounding boxes and circularity using multithreading.
    Adds 'bounding_box' and 'circularity' columns to the original DataFrame in place.
    Saves the resulting DataFrame to disk and loads it if it already exists.

    Parameters:
        images_folder (str): Path to the folder containing retina images.
        df (pd.DataFrame): A DataFrame with an 'image' column containing image names without extensions.
        max_workers (int, optional): The maximum number of threads to use in the pool. Defaults to CPU count.

    Returns:
        None: The original DataFrame `df` is updated in place with new columns.
    """

    # Define the path to save/load the DataFrame
    save_path = os.path.join(images_folder, 'retina_df.pkl')

    # Check if the DataFrame already exists
    if os.path.exists(save_path):
        print(f"Loading existing DataFrame from {save_path}")
        try:
            retina_df = pd.read_pickle(save_path)
            print("DataFrame loaded successfully.")
        except Exception as e:
            print(f"Failed to load DataFrame. Reason: {e}")
            retina_df = None
    else:
        retina_df = None

    # If DataFrame is not loaded, process the images
    if retina_df is None:
        def process_image(image_name, images_folder):
            """
            Processes a single retina image to extract required metrics.

            Parameters:
                image_name (str): Name of the image without extension.
                images_folder (str): Path to the folder containing retina images.

            Returns:
                dict or None: A dictionary with extracted metrics or None if processing fails.
            """
            image_path = os.path.join(images_folder, f"{image_name}.jpeg")
            image = cv2.imread(image_path, cv2.IMREAD_COLOR)

            if image is None:
                # Image not found or cannot be read
                print(f"Image not found or cannot be read: {image_path}")
                return None

            try:
                # Convert to grayscale and threshold to create a binary mask
                gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
                _, thresh = cv2.threshold(gray, 10, 255, cv2.THRESH_BINARY)

                # Detect contours and validate their presence
                contours, _ = cv2.findContours(thresh, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
                if not contours:
                    print(f"No contours found in image: {image_name}")
                    return None

                # Find the largest contour (assuming it's the retina)
                largest_contour = max(contours, key=cv2.contourArea)
                M = cv2.moments(largest_contour)

                # Calculate the center of the contour
                if M["m00"] != 0:
                    cX = int(M["m10"] / M["m00"])
                    cY = int(M["m01"] / M["m00"])
                else:
                    cX, cY = 0, 0

                # Normalize center coordinates by image size
                height, width = image.shape[:2]
                normalized_cX = cX / width
                normalized_cY = cY / height

                # Calculate bounding box
                x, y, w, h = cv2.boundingRect(largest_contour)

                # Convex Hull for a better approximation
                hull = cv2.convexHull(largest_contour)
                hull_area = cv2.contourArea(hull)
                hull_perimeter = cv2.arcLength(hull, True)

                # Refined Circularity using Convex Hull
                circularity = (4 * np.pi * hull_area) / (hull_perimeter ** 2) if hull_perimeter > 0 else 0

                # Return the result as a dictionary
                return {
                    "image": image_name,
                    "center_x": normalized_cX,
                    "center_y": normalized_cY,
                    "bounding_box": (x, y, w, h),
                    "circularity": circularity
                }

            except Exception as e:
                # Handle any unexpected exceptions during processing
                print(f"Error processing image {image_name}: {e}")
                return None

        retina_info = []

        # Initialize ThreadPoolExecutor
        with ThreadPoolExecutor(max_workers=max_workers) as executor:
            # Submit all image processing tasks to the thread pool
            future_to_image = {executor.submit(process_image, row['image'], images_folder): row['image']
                               for _, row in df.iterrows()}

            # Use tqdm to display progress
            for future in tqdm(as_completed(future_to_image), total=len(future_to_image), desc="Processing images", unit="image"):
                result = future.result()
                if result is not None:
                    retina_info.append(result)

        # Convert retina_info to a DataFrame
        retina_df = pd.DataFrame(retina_info)

        if retina_df.empty:
            print("No valid retina information was extracted.")
        else:
            # Save the DataFrame to disk
            try:
                retina_df.to_pickle(save_path)
                print(f"DataFrame saved to {save_path}")
            except Exception as e:
                print(f"Failed to save DataFrame. Reason: {e}")

    else:
        print("Proceeding with loaded DataFrame.")

    # --- Add 'bounding_box' and 'circularity' to the original DataFrame ---
    if retina_df is not None and not retina_df.empty:
        # Create a mapping from 'image' to 'bounding_box' and 'circularity'
        bounding_box_mapping = retina_df.set_index('image')['bounding_box'].to_dict()
        circularity_mapping = retina_df.set_index('image')['circularity'].to_dict()

        # Add 'bounding_box' and 'circularity' columns to 'df' in place
        df['bounding_box'] = df['image'].map(bounding_box_mapping)
        df['circularity'] = df['image'].map(circularity_mapping)

        print("Added 'bounding_box' and 'circularity' columns to the DataFrame.")
    else:
        print("No data available to add 'bounding_box' and 'circularity' columns.")

    # --- Create a Plot for Retina Circularity ---
    if retina_df is not None and not retina_df.empty:
        plt.figure(figsize=(10, 10))
        scatter = plt.scatter(retina_df['center_x'], retina_df['center_y'],
                              c=retina_df['circularity'], cmap='jet',
                              s=100, alpha=0.9, edgecolors='k')
        plt.title("Refined Retina Circularity", fontsize=16)
        plt.xlabel("Normalized Center X", fontsize=14)
        plt.ylabel("Normalized Center Y", fontsize=14)
        plt.grid(True, alpha=0.7)
        plt.xlim(0, 1)
        plt.ylim(0, 1)
        plt.xticks(np.arange(0, 1.1, 0.1))
        plt.yticks(np.arange(0, 1.1, 0.1))
        cbar = plt.colorbar(scatter)  # Add a color bar for circularity
        cbar.set_label("Circularity (1 = Perfect Circle)", fontsize=12)
        cbar.set_ticks(np.arange(0, 1.1, 0.1))
        plt.tight_layout()
        plt.show()
    else:
        print("No data available for plotting.")


    # Violin Plot for Circularity
    plt.figure(figsize=(10, 4))
    sns.boxplot(x=retina_df["circularity"], color="green")
    plt.title("Box Plot of Circularity")
    plt.ylabel("Circularity Score")
    plt.xticks(np.arange(0, 1.05, 0.05))
    plt.tight_layout()
    plt.show()



    # Define circularity values we want to display
    target_circularities = [1, 0.99, 0.95, 0.9, 0.85]

    # Find the closest matching images
    selected_images = []
    for target in target_circularities:
        closest_match = retina_df.iloc[(retina_df["circularity"] - target).abs().argsort()[:1]]
        selected_images.append(closest_match)

    # Flatten the list and remove duplicates
    selected_images = pd.concat(selected_images).drop_duplicates(subset=["image"])

    # Plot the images in a subplot
    num_images = len(selected_images)
    fig, axes = plt.subplots(1, num_images, figsize=(3 * num_images, 3))

    for ax, (_, row) in zip(axes, selected_images.iterrows()):
        image_path = os.path.join(images_folder, f"{row['image']}.jpeg")
        img = cv2.imread(image_path)
        img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)  # Convert BGR to RGB for correct display

        ax.imshow(img)
        ax.set_title(f"Circularity: {row['circularity']:.2f}")
        ax.axis("off")

    plt.tight_layout()
    plt.show()


analyze_retina_images(images_path, df)
```

```python
def analyze_brightness_contrast(df, images_folder, max_workers=os.cpu_count() if os.cpu_count() else 4):
    """
    Analyze brightness and contrast of all images within the specified folder,
    regardless of their exposure categories, utilizing multithreading for efficiency.

    Parameters:
        df (pd.DataFrame): DataFrame containing at least the following columns:
            - 'image': Image filename without extension.
            - 'level': Integer class label (e.g., disease levels).
        images_folder (str): Path to the root folder containing all exposure categories as subdirectories.
        sample_size (int): Number of samples to analyze per class label.
        random_state (int): Seed for reproducibility in sampling.
        max_workers (int): Number of threads to use in the thread pool (default is 8).

    Returns:
        None: Displays histograms of brightness and contrast distributions by class.
    """

    # Initialize lists to store metrics
    brightness_list = []
    contrast_list = []
    label_list = []
    image_list = []

    # Sample data from each class label
    sampled_data = df.copy()

    def process_image(row):
        """
        Process a single image to compute its brightness and contrast.

        Parameters:
            row (pd.Series): A row from the DataFrame containing 'image' and 'level'.

        Returns:
            tuple: (brightness, contrast, level, image_name) or None if processing fails.
        """
        image_name = row['image']
        level = row['level']
        image_path = os.path.join(images_folder, image_name + ".jpeg")

        # Check if the image exists
        if not os.path.exists(image_path):
            # Image not found; skip processing
            return None

        # Read the image in grayscale
        image = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)

        if image is None:
            # Failed to read the image; skip processing
            return None

        # Compute brightness and contrast
        brightness = np.mean(image)
        contrast = np.std(image)

        return (brightness, contrast, level, image_name)

    # Use ThreadPoolExecutor for multithreading
    with ThreadPoolExecutor(max_workers=max_workers) as executor:
        # Submit all image processing tasks
        futures = {executor.submit(process_image, row): idx for idx, row in sampled_data.iterrows()}

        # Use tqdm to display progress
        for future in tqdm(as_completed(futures), total=len(futures), desc="Analyzing Brightness & Contrast", unit='images'):
            result = future.result()
            if result is not None:
                brightness, contrast, level, image_name = result
                brightness_list.append(brightness)
                contrast_list.append(contrast)
                label_list.append(level)
                image_list.append(image_name)

    # Check if any images were processed
    if not brightness_list:
        print("No images were processed. Please check the image paths and DataFrame contents.")
        return

    # Visualization: Brightness and Contrast Distribution by Class
    unique_labels = sorted(set(label_list))
    num_labels = len(unique_labels)

    # Determine subplot grid size based on the number of classes
    cols = min(5, num_labels)  # Maximum 5 columns
    rows = 2  # Fixed to 2 rows: one for brightness, one for contrast

    fig, axes = plt.subplots(rows, cols, figsize=(5 * cols, 10))
    fig.suptitle("Brightness and Contrast Distribution by Class", fontsize=16)

    # If there's only one column, axes may not be a 2D array
    if cols == 1:
        axes = np.expand_dims(axes, axis=1)

    for i, label in enumerate(unique_labels):
        # Select the appropriate subplot for brightness and contrast
        ax_brightness = axes[0, i]
        ax_contrast = axes[1, i]

        # Filter data for the current label
        brightness_data = [b for b, l in zip(brightness_list, label_list) if l == label]
        contrast_data = [c for c, l in zip(contrast_list, label_list) if l == label]

        # Plot Brightness Histogram
        ax_brightness.hist(brightness_data, bins=25, alpha=0.7, color='skyblue', edgecolor='black')
        ax_brightness.set_title(f"Class {label} Brightness")
        ax_brightness.set_xlabel("Brightness")
        ax_brightness.set_ylabel("Frequency")
        ax_brightness.grid(True, linestyle='--', alpha=0.7)

        # Calculate statistics for brightness
        mean_b = np.mean(brightness_data)
        median_b = np.median(brightness_data)
        mode_result_b = stats.mode(brightness_data, nan_policy='omit')
        # Handle mode based on scipy version
        if np.isscalar(mode_result_b.mode):
            mode_b = mode_result_b.mode if mode_result_b.count > 0 else mean_b
        else:
            mode_b = mode_result_b.mode[0] if mode_result_b.count[0] > 0 else mean_b

        # Plot vertical lines for brightness
        ax_brightness.axvline(mean_b, color='red', linestyle='dashed', linewidth=1.5, label='Mean')
        ax_brightness.axvline(median_b, color='green', linestyle='dotted', linewidth=1.5, label='Median')
        ax_brightness.axvline(mode_b, color='orange', linestyle='dashdot', linewidth=1.5, label='Mode')
        ax_brightness.legend()

        # Plot Contrast Histogram
        ax_contrast.hist(contrast_data, bins=25, alpha=0.7, color='lightgreen', edgecolor='black')
        ax_contrast.set_title(f"Class {label} Contrast")
        ax_contrast.set_xlabel("Contrast")
        ax_contrast.set_ylabel("Frequency")
        ax_contrast.grid(True, linestyle='--', alpha=0.7)

        # Calculate statistics for contrast
        mean_c = np.mean(contrast_data)
        median_c = np.median(contrast_data)
        mode_result_c = stats.mode(contrast_data, nan_policy='omit')

        # Handle mode based on scipy version
        if np.isscalar(mode_result_c.mode):
            mode_c = mode_result_c.mode if mode_result_c.count > 0 else mean_c
        else:
            mode_c = mode_result_c.mode[0] if mode_result_c.count[0] > 0 else mean_c

        # Plot vertical lines for contrast
        ax_contrast.axvline(mean_c, color='red', linestyle='dashed', linewidth=1.5, label='Mean')
        ax_contrast.axvline(median_c, color='green', linestyle='dotted', linewidth=1.5, label='Median')
        ax_contrast.axvline(mode_c, color='orange', linestyle='dashdot', linewidth=1.5, label='Mode')
        ax_contrast.legend()

    # Hide any unused subplots
    for j in range(i + 1, cols):
        ax_brightness = axes[0, j]
        ax_contrast = axes[1, j]
        ax_brightness.axis('off')
        ax_contrast.axis('off')

    plt.tight_layout(rect=[0, 0.03, 1, 0.95])
    plt.show()

    # Create a DataFrame from the existing lists
    df_metrics = pd.DataFrame({
        'Brightness': brightness_list,
        'Contrast': contrast_list,
        'Class': label_list
    })

    # Reshape the DataFrame for easier plotting (one column for values, one for metric type)
    df_melted = pd.melt(df_metrics, id_vars='Class', value_vars=['Brightness', 'Contrast'],
                        var_name='Metric', value_name='Value')

    # Plot a grouped boxplot
    plt.figure(figsize=(20, 6))
    sns.boxplot(x='Class', y='Value', hue='Metric', data=df_melted)
    plt.title("Boxplot of Brightness and Contrast by Class")
    plt.xlabel("Class")
    plt.ylabel("Value")
    plt.legend(title="Metric")
    plt.tight_layout()
    plt.show()


analyze_brightness_contrast(df, images_path)
```

Now lets turn our attention to colors.

```python
def analyze_global_pixel_intensity(df, images_folder, bins=256, max_workers=os.cpu_count() if os.cpu_count() else 4):
    """
    Analyzes global pixel intensity, plots prototype histograms for each class, and summarizes channel statistics.

    Parameters
    ----------
    df : pd.DataFrame
        DataFrame with two columns: 'image' (image filename without extension) and 'level' (class label).
    images_folder : str
        Path to the folder containing {image}.jpeg files.
    bins : int, optional
        Number of bins in the stored histogram.
    max_workers : int, optional
        Maximum number of threads to use for parallel processing.
    """

    def compute_image_stats(args):
        """Helper function to compute statistics for a single image"""
        image_path, image_name, level = args
        img_bgr = cv2.imread(image_path)

        if img_bgr is None:
            return None

        # Split channels in BGR order
        b_channel, g_channel, r_channel = cv2.split(img_bgr)

        # Compute mean and std per channel
        stats = {
            'label': level,
            'image': image_name,
            'mean_b': b_channel.mean(),
            'std_b': b_channel.std(),
            'mean_g': g_channel.mean(),
            'std_g': g_channel.std(),
            'mean_r': r_channel.mean(),
            'std_r': r_channel.std()
        }

        # Compute RGB histogram
        hist_b = cv2.calcHist([img_bgr], [0], None, [256], [0, 256]).ravel()
        hist_g = cv2.calcHist([img_bgr], [1], None, [256], [0, 256]).ravel()
        hist_r = cv2.calcHist([img_bgr], [2], None, [256], [0, 256]).ravel()

        stats['rgb_hist'] = {'b': hist_b, 'g': hist_g, 'r': hist_r}

        return stats

    # Prepare arguments for parallel processing
    process_args = [(os.path.join(images_folder, f"{row['image']}.jpeg"), row['image'], row['level'])
                    for _, row in df.iterrows()]

    # Initialize containers
    stats_records = []
    prototype_hist = {cl: {'b': np.zeros(bins), 'g': np.zeros(bins), 'r': np.zeros(bins), 'count': 0}
                      for cl in sorted(df['level'].unique())}

    # Compute statistics in parallel
    with ThreadPoolExecutor(max_workers=max_workers) as executor:
        futures = [executor.submit(compute_image_stats, args) for args in process_args]

        for future in tqdm(as_completed(futures), total=len(futures)):
            result = future.result()
            if result is not None:
                stats_records.append({k: v for k, v in result.items() if k != 'rgb_hist'})

                # Update prototype histograms
                cl = result['label']
                rgb_hist = result['rgb_hist']
                for channel in ['b', 'g', 'r']:
                    prototype_hist[cl][channel] += rgb_hist[channel]
                prototype_hist[cl]['count'] += 1

    # Create statistics DataFrame
    stats_df = pd.DataFrame(stats_records)

    # Aggregate by class
    channel_summary = stats_df.groupby('label').agg({'mean_b': 'mean', 'mean_g': 'mean', 'mean_r': 'mean',
                                                     'std_b': 'mean', 'std_g': 'mean', 'std_r': 'mean'}).reset_index()

    print("\n=== Per-Class Channel Summary (Mean of Means) ===")
    print(channel_summary)

    # Plot in a single figure with 2 rows and 3 columns
    fig, axes = plt.subplots(2, 3, figsize=(18, 12))
    axes = axes.flatten()

    # Define color mappings for channels
    channel_palette = {'mean_b': 'blue',
                       'mean_g': 'green',
                       'mean_r': 'red'}

    channel_colors = {'b': 'blue',
                      'g': 'green',
                      'r': 'red'}

    # Plot 1: Bar plot of channel means
    wide_df = channel_summary.melt(
        id_vars=['label'],
        value_vars=['mean_b', 'mean_g', 'mean_r'],
        var_name='channel_stat',
        value_name='value'
    )

    # Bar Plot
    sns.barplot(
        x='label',
        y='value',
        hue='channel_stat',
        data=wide_df,
        ax=axes[0],
        palette=channel_palette  # Apply the custom palette
    )
    axes[0].set_title("Average Mean Channel Intensity by Class")
    axes[0].set_xlabel("Class")
    axes[0].set_ylabel("Mean Intensity")
    axes[0].legend(title="Channel")
    axes[0].grid(alpha=0.5)

    # Plot 2-6: Class prototype histograms
    unique_classes = sorted(df['level'].unique())
    for idx, cl in enumerate(unique_classes):
        if idx + 1 >= len(axes):  # Skip if there are more classes than available subplots
            break

        count = prototype_hist[cl]['count']
        if count == 0:
            axes[idx + 1].axis('off')
            continue

        ax = axes[idx + 1]
        for channel in ['b', 'g', 'r']:
            avg_hist = prototype_hist[cl][channel]
            ax.plot(range(bins),
                    avg_hist,
                    label=f"{channel.upper()} Channel",
                    color=channel_colors[channel])  # Assign the correct color

        ax.set_title(f"Class {cl} (Avg of {count} Images)")
        ax.set_xlabel("Intensity")
        ax.set_ylabel("Frequency (Log Scale)")
        ax.set_yscale("log")  # Apply log scale
        ax.legend()
        ax.grid(alpha=0.5)

    # Hide unused subplots if there are fewer classes than grid cells
    for idx in range(len(unique_classes) + 1, len(axes)):
        axes[idx].axis('off')

    plt.tight_layout(rect=[0, 0, 1, 0.95])
    fig.suptitle("Channel Statistics and Prototype Histograms", fontsize=16)
    plt.show()

analyze_global_pixel_intensity(df, images_path)
```

The Red Channel (mean_r) consistently exhibits the highest mean intensity across all classes, suggesting that the images have a reddish tint, which is common in retinal imaging due to the natural coloration of blood vessels and retinal tissue. The Blue and Green channels are less pronounced, aligning with the expected biological characteristics of the retina.
```python
def process_images(df, images_folder, max_workers=os.cpu_count() if os.cpu_count() else 4):
    """
    Processes images to compute their histograms, adds a 'histogram' column to the DataFrame in place,
    and plots the aggregated histogram with highlighted high variability bins.

    Parameters:
        df (pd.DataFrame): DataFrame containing 'image' and 'level' columns.
        images_folder (str): Path to the folder containing images.
        max_workers (int, optional): Number of threads for parallel processing. Defaults to CPU count.

    Returns:
        None: The original DataFrame `df` is updated in place with a new 'histogram' column.
    """

    def process_single_row(row):
        """
        Processes a single image to compute its histogram.

        Parameters:
            row (pd.Series): Row containing 'image' and 'level'.

        Returns:
            dict: Dictionary with 'image', 'level', and 'histogram' or error info.
        """
        image_name = row["image"]
        label = row["level"]
        image_path = os.path.join(images_folder, f"{image_name}.jpeg")

        result = {
            "image": image_name,
            "level": label,
            "histogram": None
        }

        if not os.path.isfile(image_path):
            # Image not found
            print(f"Warning: Image not found: {image_path}")
            return result

        try:
            # Load the image in grayscale
            gray_image = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)

            if gray_image is None:
                # Failed to read image
                print(f"Warning: Failed to read image: {image_path}")
                return result

            # Compute histogram using OpenCV's calcHist for efficiency
            hist = cv2.calcHist([gray_image], [0], None, [256], [0, 256]).flatten()
            hist_list = hist.tolist()

            result["histogram"] = hist_list

            return result

        except Exception as e:
            print(f"Error processing {image_name}: {e}")
            return result

    # Initialize lists to collect results
    all_results = []
    all_hists = []

    # Process images in parallel using ThreadPoolExecutor
    with ThreadPoolExecutor(max_workers=max_workers) as executor:
        # Submit all image processing tasks to the thread pool
        futures = [executor.submit(process_single_row, row) for _, row in df.iterrows()]

        # Use tqdm to display progress
        for future in tqdm(as_completed(futures), total=len(futures), desc="Processing Images", unit="image"):
            res = future.result()
            all_results.append(res)
            if res["histogram"] is not None:
                all_hists.append(res["histogram"])

    # Convert all_results to a DataFrame
    results_df = pd.DataFrame(all_results)

    if results_df.empty:
        print("No histograms were computed.")
        return

    # Create a mapping from 'image' to 'histogram'
    histogram_mapping = results_df.set_index('image')['histogram'].to_dict()

    # Add 'histogram' column to the original df in place
    df['histogram'] = df['image'].map(histogram_mapping)

    print("Added 'histogram' column to the DataFrame.")

    # Aggregate histograms for plotting
    if all_hists:
        # Convert list of histograms to a numpy array
        valid_hists = np.array(all_hists, dtype=float)

        # Compute the mean and standard deviation across all histograms
        aggregated_histogram = np.mean(valid_hists, axis=0)
        std_histogram = np.std(valid_hists, axis=0)

        # Identify bins with high variability (e.g., where std > 3.5 * mean for that bin)
        # To avoid division by zero, ensure that mean is not zero
        high_variability_bins = np.where(aggregated_histogram > 0)[0]
        high_variability_bins = np.where(std_histogram > 3.5 * aggregated_histogram)[0]

        # Plot the aggregated histogram with ±std and highlight high variability bins
        intensities = np.arange(256)

        # Determine symmetric limit for log scale
        max_val = np.max(aggregated_histogram + std_histogram)
        symmetric_limit = 10 ** np.ceil(np.log10(max_val)) if max_val > 0 else 1

        plt.figure(figsize=(16, 7))

        # Plot the mean histogram as bars
        plt.bar(intensities, aggregated_histogram, color="gray", alpha=0.7, label="Mean Histogram")

        # Plot ±std as a shaded region
        plt.fill_between(intensities, aggregated_histogram - std_histogram, aggregated_histogram + std_histogram,
                         color="blue", alpha=0.3, label="Std Dev Range")

        # Highlight bins with high variability
        plt.bar(intensities[high_variability_bins],
                aggregated_histogram[high_variability_bins],
                color="red",
                alpha=0.8,
                label="High Variability Bins")

        # Add vertical lines for key statistics
        total_pixels = np.sum(aggregated_histogram)
        mean_val = np.sum(aggregated_histogram * intensities) / total_pixels
        cdf = np.cumsum(aggregated_histogram) / total_pixels
        median_val = np.interp(0.5, cdf, intensities)
        mode_val = intensities[np.argmax(aggregated_histogram)]

        plt.axvline(mean_val, color="red", linestyle="--", linewidth=1.5, label=f"Mean: {mean_val:.2f}")
        plt.axvline(median_val, color="green", linestyle="--", linewidth=1.5, label=f"Median: {median_val:.2f}")
        plt.axvline(mode_val, color="orange", linestyle="--", linewidth=1.5, label=f"Mode: {mode_val:.0f}")

        # Plot settings
        plt.title("Aggregated Pixel Intensity Histogram with High Variability Bins", fontsize=16)
        plt.xlabel("Pixel Intensity", fontsize=14)
        plt.ylabel("Frequency", fontsize=14)
        plt.legend(fontsize=12)
        plt.grid(alpha=0.8)
        plt.xticks(ticks=np.arange(0, 256 + 8, 8))  # Ticks from 0 to 256 with a step of 8
        plt.yscale("log")  # Symmetric log scale with a linear threshold at ±1
        plt.ylim(1, symmetric_limit)  # Ensure symmetric y-axis limits
        plt.tight_layout()
        plt.show()
    else:
        print("No valid histograms were extracted for plotting.")

process_images(df, images_path)
```

```python
def extract_classic_features_components(image_path):
    """
    Reads an image, resizes it to 256x256, and computes several classic feature visualizations.

    Returns a dictionary with:
      - 'resized_image': the resized image,
      - 'hog_image': visualization of HOG features,
      - 'lbp_image': the LBP output image,
      - 'gabor_example': an example Gabor filter response (frequency=0.2, theta=0),
      - 'log_image': visualization of the Laplacian of Gaussian (LoG),
      - 'features': concatenated feature vector including all extracted features,
      - 'glcm_features': Gray-Level Co-occurrence Matrix (GLCM) features,
      - 'color_hist': normalized color histogram (empty if grayscale),
      - 'gabor_features': extracted Gabor features,
      - 'haralick_features': Haralick texture features.
    """
    # Read and resize image to 256x256
    image = io.imread(image_path)
    image = transform.resize(image, (256, 256), anti_aliasing=True)

    # Convert to grayscale if necessary
    gray = color.rgb2gray(image) if image.ndim == 3 else image
    gray_int = img_as_ubyte(gray)

    # --- HOG Features ---
    hog_features, hog_image = hog(
        gray,
        pixels_per_cell=(16, 16),
        cells_per_block=(2, 2),
        visualize=True,
        feature_vector=True
    )

    # --- LBP Features ---
    radius = 1
    n_points = 8 * radius
    lbp = local_binary_pattern(gray_int, n_points, radius, method='uniform')
    lbp_image = lbp.copy()

    # --- GLCM Features ---
    gray_u8 = img_as_ubyte(gray)
    glcm = graycomatrix(gray_u8, distances=[1], angles=[0], symmetric=True, normed=True)
    glcm_features = np.array([
        graycoprops(glcm, 'contrast')[0, 0],
        graycoprops(glcm, 'dissimilarity')[0, 0],
        graycoprops(glcm, 'homogeneity')[0, 0],
        graycoprops(glcm, 'energy')[0, 0],
        graycoprops(glcm, 'correlation')[0, 0],
        graycoprops(glcm, 'ASM')[0, 0]
    ])

    # --- Color Histograms (for color images only) ---
    if image.ndim == 3:
        hist_r, _ = np.histogram(image[:, :, 0], bins=32, range=(0, 1))
        hist_g, _ = np.histogram(image[:, :, 1], bins=32, range=(0, 1))
        hist_b, _ = np.histogram(image[:, :, 2], bins=32, range=(0, 1))
        color_hist = np.concatenate([hist_r, hist_g, hist_b]).astype("float")
        color_hist /= (color_hist.sum() + 1e-6)
    else:
        color_hist = np.array([])

    # --- Gabor Filter Features ---
    gabor_feats = []
    frequencies = [0.1, 0.2, 0.3]
    thetas = [0, np.pi/4, np.pi/2, 3*np.pi/4]
    gabor_example = None
    for frequency in frequencies:
        for theta in thetas:
            filt_real, _ = gabor(gray, frequency=frequency, theta=theta)
            gabor_feats.append(filt_real.mean())
            gabor_feats.append(filt_real.var())
            if frequency == 0.2 and theta == 0:
                gabor_example = filt_real.copy()
    gabor_features = np.array(gabor_feats)

    # --- Haralick Texture Features ---
    haralick_features = mahotas.features.haralick(gray_int).mean(axis=0)

    # --- Laplacian of Gaussian (LoG) ---
    log_image = laplace(gaussian(gray, sigma=1))

    # Combine all features into one vector
    features = np.concatenate([
        hog_features,
        glcm_features,
        gabor_features,
        haralick_features,
        color_hist
    ])

    return {
        'features': features,
        'resized_image': image,
        'hog_image': hog_image,
        'lbp_image': lbp_image,
        'gabor_example': gabor_example,
        'log_image': log_image,
        'glcm_features': glcm_features,
        'color_hist': color_hist,
        'gabor_features': gabor_features,
        'haralick_features': haralick_features
    }


def visualize_filters():
    """
    Samples one image per severity level from a CSV (with 'image' and 'level' columns),
    extracts classic features, and displays the original image along with HOG, LBP,
    Gabor, Sobel, Sato vesselness, and LoG visualizations.
    """
    # Sample one image per level from the dataframe
    sample_df = df.groupby('level', group_keys=False).apply(
        lambda x: x.sample(n=1, random_state=37)
    ).reset_index(drop=True)

    sample_files = sample_df['image'].astype(str).tolist()
    severity_levels = sample_df['level'].tolist()

    # Extract classic feature components for each sample image
    classic_components_list = []
    for file in sample_files:
        image_path = os.path.join(images_path, f"{file}.jpeg")
        comp = extract_classic_features_components(image_path)
        classic_components_list.append(comp)

    n_examples = len(classic_components_list)

    # Update: Add one more column for LoG visualization (now 7 columns in total)
    fig, axes = plt.subplots(n_examples, 7, figsize=(21, 3 * n_examples))
    fig.suptitle("Visualization of Classic Filters per Severity Level", fontsize=16)

    for i, comp in enumerate(classic_components_list):
        orig_img = comp['resized_image']
        hog_img = comp['hog_image']
        lbp_img = comp['lbp_image']
        gabor_img = comp['gabor_example']
        log_img = comp['log_image']

        # Convert the resized image to grayscale (if needed)
        gray_img = color.rgb2gray(orig_img) if orig_img.ndim == 3 else orig_img

        # Compute Sobel edge detection
        sobel_img = sobel(gray_img)

        # --- Sato Vesselness Detector ---
        t0, t1 = threshold_multiotsu(gray_img, classes=3)
        mask = gray_img > t0
        sato_img = sato(gray_img, sigmas=range(1, 10)) * mask

        # List of image-title pairs
        images_titles = [
            (orig_img, f"Original (Level {severity_levels[i]})"),
            (hog_img, "HOG"),
            (lbp_img, "LBP"),
            (gabor_img, "Gabor"),
            (sobel_img, "Sobel"),
            (sato_img, "Sato Vesselness"),
            (log_img, "LoG (Laplacian of Gaussian)")
        ]

        # Loop over each image-title pair and display them in the proper subplot
        for j, (img, title) in enumerate(images_titles):
            # Use 'magma' colormap for Sato Vesselness, 'gray' for all others except the original image
            if j == 0:
                cmap = None
            elif title == "Sato Vesselness":
                cmap = "magma"
            else:
                cmap = "gray"

            axes[i, j].imshow(img, cmap=cmap)
            axes[i, j].set_title(title)
            axes[i, j].axis("off")

    plt.tight_layout(rect=[0, 0.03, 1, 0.95])
    plt.show()


visualize_filters()
```

By leveraging both traditional feature extraction techniques and deep learning embeddings, I aim to determine whether a hybrid approach can improve clustering performance. I'll extract key features from methods like LBP, Sato Vesselness, and LoG while integrating them with deep learning-based representations from *EfficientNetV2*.
```python
def extract_features(images, images_path):
    """
    Extract both deep learned features and classic features from the given images.
    Returns two numpy arrays: deep_features and classic_features.
    """
    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    # Load EfficientNet_v2_m with pretrained weights
    model = efficientnet_v2_m(weights=EfficientNet_V2_M_Weights.DEFAULT)
    model.to(device).eval()

    deep_features_list = []
    classic_features_list = []

    # Use 384x384 as the input size for EfficientNet_v2_m.
    transform = transforms.Compose([
        transforms.Resize((384, 384)),
        transforms.ToTensor(),
        transforms.Normalize(mean=[0.485, 0.456, 0.406],
                             std=[0.229, 0.224, 0.225])
    ])

    for img_name in tqdm(images, desc="Extracting Deep & Classic Features"):
        img_path = os.path.join(images_path, f"{img_name}.jpeg")

        # Deep learned features extraction:
        image = Image.open(img_path).convert("RGB")
        image_tensor = transform(image).unsqueeze(0).to(device)
        with torch.no_grad():
            efficient_features = None
            def hook_fn(module, input, output):
                nonlocal efficient_features
                efficient_features = output
            hook = model.features[-1].register_forward_hook(hook_fn)
            _ = model(image_tensor)
            hook.remove()
            # Global average pooling over spatial dimensions:
            efficient_features = efficient_features.mean(dim=[2, 3]).cpu().numpy().flatten()
        deep_features_list.append(efficient_features)

        # Classic features extraction:
        classic_dict = extract_classic_features_components(img_path)
        classic_features = classic_dict['features']
        classic_features_list.append(classic_features)

    return np.array(deep_features_list), np.array(classic_features_list)


def reduce_and_plot(features, labels, plot_title=""):
    """
    Perform dimensionality reduction using PCA, t-SNE, UMAP, and LDA (for supervised reduction),
    run clustering (K-means and HDBSCAN) on the 2D and 3D embeddings, and plot the results.
    """
    # Normalize features
    scaler = StandardScaler()
    features = scaler.fit_transform(features)

    methods = ["PCA", "t-SNE", "UMAP", "LDA"]
    n_classes = len(np.unique(labels))  # Number of unique ground truth labels

    reducers_2d = [
        PCA(n_components=2),
        TSNE(n_components=2, random_state=random_state, perplexity=25),
        umap.UMAP(n_components=2, random_state=random_state, n_neighbors=25, min_dist=0.01),
        LDA(n_components=min(n_classes-1, 2))  # LDA max components = n_classes-1
    ]

    reducers_3d = [
        PCA(n_components=3),
        TSNE(n_components=3, random_state=random_state, perplexity=25),
        umap.UMAP(n_components=3, random_state=random_state, n_neighbors=25, min_dist=0.01),
        LDA(n_components=min(n_classes-1, 3))  # LDA max components = n_classes-1
    ]

    # Create figure with 2 rows: top row for 2D plots, bottom row for 3D plots
    fig = plt.figure(figsize=(26, 16))
    if plot_title:
        fig.suptitle(plot_title, fontsize=20)
    axes_2d = [fig.add_subplot(2, 4, i+1) for i in range(4)]
    axes_3d = [fig.add_subplot(2, 4, i+5, projection='3d') for i in range(4)]

    n_clusters = len(np.unique(labels))

    for i, method in enumerate(methods):
        # For LDA, use labels in fit_transform
        if method == "LDA":
            embed_2d = reducers_2d[i].fit_transform(features, labels)
            embed_3d = reducers_3d[i].fit_transform(features, labels)
        else:
            embed_2d = reducers_2d[i].fit_transform(features)
            embed_3d = reducers_3d[i].fit_transform(features)

        sil_score_2d = silhouette_score(embed_2d, labels)
        sil_score_3d = silhouette_score(embed_3d, labels)

        # Clustering on 2D embedding
        kmeans = KMeans(n_clusters=n_clusters, random_state=random_state)
        kmeans_labels = kmeans.fit_predict(embed_2d)
        try:
            sil_kmeans = silhouette_score(embed_2d, kmeans_labels)
            dbi_kmeans = davies_bouldin_score(embed_2d, kmeans_labels)
            chi_kmeans = calinski_harabasz_score(embed_2d, kmeans_labels)
        except:
            sil_kmeans = dbi_kmeans = chi_kmeans = np.nan

        clusterer = hdbscan.HDBSCAN(min_cluster_size=5)
        hdbscan_labels = clusterer.fit_predict(embed_2d)
        if len(set(hdbscan_labels) - {-1}) > 1:
            try:
                sil_hdbscan = silhouette_score(embed_2d, hdbscan_labels)
                dbi_hdbscan = davies_bouldin_score(embed_2d, hdbscan_labels)
                chi_hdbscan = calinski_harabasz_score(embed_2d, hdbscan_labels)
            except:
                sil_hdbscan = dbi_hdbscan = chi_hdbscan = np.nan
        else:
            sil_hdbscan = dbi_hdbscan = chi_hdbscan = np.nan

        # Clustering on 3D embedding
        kmeans_3d = KMeans(n_clusters=n_clusters, random_state=random_state)
        kmeans_labels_3d = kmeans_3d.fit_predict(embed_3d)
        try:
            sil_kmeans_3d = silhouette_score(embed_3d, kmeans_labels_3d)
            dbi_kmeans_3d = davies_bouldin_score(embed_3d, kmeans_labels_3d)
            chi_kmeans_3d = calinski_harabasz_score(embed_3d, kmeans_labels_3d)
        except:
            sil_kmeans_3d = dbi_kmeans_3d = chi_kmeans_3d = np.nan

        clusterer_3d = hdbscan.HDBSCAN(min_cluster_size=5)
        hdbscan_labels_3d = clusterer_3d.fit_predict(embed_3d)
        if len(set(hdbscan_labels_3d) - {-1}) > 1:
            try:
                sil_hdbscan_3d = silhouette_score(embed_3d, hdbscan_labels_3d)
                dbi_hdbscan_3d = davies_bouldin_score(embed_3d, hdbscan_labels_3d)
                chi_hdbscan_3d = calinski_harabasz_score(embed_3d, hdbscan_labels_3d)
            except:
                sil_hdbscan_3d = dbi_hdbscan_3d = chi_hdbscan_3d = np.nan
        else:
            sil_hdbscan_3d = dbi_hdbscan_3d = chi_hdbscan_3d = np.nan

        # Plotting the results
        ax2d = axes_2d[i]
        ax3d = axes_3d[i]
        for cls in np.unique(labels):
            idx = labels == cls
            ax2d.scatter(embed_2d[idx, 0], embed_2d[idx, 1], label=str(cls), alpha=0.7)
            ax3d.scatter(embed_3d[idx, 0], embed_3d[idx, 1], embed_3d[idx, 2], label=str(cls), alpha=0.7)

        title_2d = (f"{method} 2D Visualization\n"
                    f"GT Silhouette: {sil_score_2d:.3f}\n"
                    f"K-means -> Sil: {sil_kmeans:.3f}, DBI: {dbi_kmeans:.3f}, CHI: {chi_kmeans:.1f}\n"
                    f"HDBSCAN -> Sil: {sil_hdbscan:.3f}, DBI: {dbi_hdbscan:.3f}, CHI: {chi_hdbscan:.1f}")
        ax2d.set_title(title_2d)
        ax2d.set_xlabel("Component 1")
        ax2d.set_ylabel("Component 2")
        ax2d.legend()

        title_3d = (f"{method} 3D Visualization\n"
                    f"GT Silhouette: {sil_score_3d:.3f}\n"
                    f"K-means -> Sil: {sil_kmeans_3d:.3f}, DBI: {dbi_kmeans_3d:.3f}, CHI: {chi_kmeans_3d:.1f}\n"
                    f"HDBSCAN -> Sil: {sil_hdbscan_3d:.3f}, DBI: {dbi_hdbscan_3d:.3f}, CHI: {chi_hdbscan_3d:.1f}")
        ax3d.set_title(title_3d)
        ax3d.set_xlabel("Component 1")
        ax3d.set_ylabel("Component 2")
        ax3d.set_zlabel("Component 3")
        ax3d.legend()

    plt.tight_layout(rect=[0, 0.03, 1, 0.95])
    plt.show()


sample_df = df.groupby('level', group_keys=False).apply(
    lambda x: x.sample(n=min(len(x), 512), random_state=random_state)).reset_index(drop=True)
images = sample_df['image'].astype(str).tolist()
labels = sample_df['level'].values

# Extract features (both deep and classic) in one pass:
deep_features, classic_features = extract_features(images, images_path)

# Compute combined features by concatenating deep and classic features:
combined_features = np.concatenate([deep_features, classic_features], axis=1)

# Plot the results for each feature type:
reduce_and_plot(deep_features, labels, plot_title="Deep Learned Features")
reduce_and_plot(classic_features, labels, plot_title="Classic Features")
reduce_and_plot(combined_features, labels, plot_title="Combined Features")
```

```python
def adaptive_square_cropping(df, images_folder, output_size=(512, 512), padding_color=(0, 0, 0),
                             max_workers=os.cpu_count() if os.cpu_count() else 4):
    """
    Perform adaptive square cropping on retinal images while maintaining aspect ratio by adding padding.
    Utilizes multithreading and saves processed images to disk to avoid redundant processing.

    Parameters:
        df (pd.DataFrame): DataFrame with 'image' (image names) and 'bounding_box' (tuple of (x, y, w, h)).
        images_folder (str): Path to the folder containing the images.
        output_size (tuple): Target size for resizing after cropping (default is 512x512).
        padding_color (tuple): RGB color for padding (default is black: (0, 0, 0)).
        max_workers (int): Number of threads to use in the thread pool.

    Returns:
        str: Path to the 'preprocessed' folder containing cropped images.
    """

    # Create output folder if it doesn’t exist
    preprocessed_folder = os.path.join(images_folder, "preprocessed")
    os.makedirs(preprocessed_folder, exist_ok=True)

    def process_image(row):
        image_name = row["image"]
        bounding_box = row["bounding_box"]  # (x, y, w, h)

        # Construct full image path
        image_path = os.path.join(images_folder, f"{image_name}.jpeg")
        save_path = os.path.join(preprocessed_folder, f"{image_name}.jpeg")

        # Check if processed image already exists
        if os.path.isfile(save_path):
            # No need to process; already saved on disk
            return

        # If processed image doesn't exist, proceed with processing
        if not os.path.isfile(image_path):
            print(f"Warning: {image_path} not found, skipping.")
            return

        # Load the original image
        image = cv2.imread(image_path)
        if image is None:
            print(f"Warning: Failed to read {image_name}, skipping.")
            return

        # Extract bounding box
        x, y, w, h = bounding_box

        # Compute the center of the retina
        cx, cy = x + w // 2, y + h // 2

        # Determine the square side length (max of width and height)
        side_length = max(w, h)

        # Define new square crop boundaries
        x1 = max(cx - side_length // 2, 0)
        y1 = max(cy - side_length // 2, 0)
        x2 = min(cx + side_length // 2, image.shape[1])
        y2 = min(cy + side_length // 2, image.shape[0])

        # Perform cropping
        cropped_image = image[y1:y2, x1:x2]

        # Determine the new shape after cropping
        h_new, w_new = cropped_image.shape[:2]

        # Create a square canvas for padding
        square_canvas = np.full((side_length, side_length, 3), padding_color, dtype=np.uint8)

        # Center the cropped image within the square canvas
        x_offset = (side_length - w_new) // 2
        y_offset = (side_length - h_new) // 2
        square_canvas[y_offset:y_offset + h_new, x_offset:x_offset + w_new] = cropped_image

        # Resize to standard size
        final_image = cv2.resize(square_canvas, output_size, interpolation=cv2.INTER_AREA)

        # Save the processed image
        cv2.imwrite(save_path, final_image)

    # Use ThreadPoolExecutor for multithreading
    with ThreadPoolExecutor(max_workers=max_workers) as executor:
        # Submit all tasks to the executor
        futures = [executor.submit(process_image, row) for _, row in df.iterrows()]

        # Use tqdm to display progress
        for _ in tqdm(as_completed(futures), total=len(futures), desc="Processing Images", unit='images'):
            pass

    return preprocessed_folder

preprocessed_folder = adaptive_square_cropping(df, images_path)
```

In the next step, we will apply Contrast Limited Adaptive Histogram Equalization (CLAHE) to enhance image contrast while preserving fine details. CLAHE is particularly beneficial in medical imaging as it improves visibility of critical features without over-amplifying noise. Retinal fundus photographs often suffer from uneven illumination, varying contrast levels, and imaging artifacts caused by different acquisition devices. By applying CLAHE, we can locally enhance contrast, allowing better differentiation of anatomical structures. This method also helps correct illumination inconsistencies, reducing the impact of uneven lighting conditions across different imaging devices. Most importantly, CLAHE makes subtle lesions more visible, which is crucial for the early diagnosis of diabetic retinopathy and other retinal diseases.

```python
def process_and_display_images(images_folder, df, output_folder=None, max_workers=os.cpu_count() if os.cpu_count() else 4):
    """
    Apply CLAHE to all images in a specified folder using multi-threading,
    save the processed images, and display example images before and after processing.

    Parameters:
        images_folder (str): Path to the folder containing input images.
        df (pd.DataFrame): DataFrame with at least 'image' and 'level' columns.
        output_folder (str, optional): Path to save processed images.
                                       If None, a 'processed' subfolder is created within images_folder.
        max_workers (int, optional): Number of threads to use. Defaults to the number of CPU cores or 4.
        num_pairs (int, optional): Number of image pairs to display. Defaults to 5.

    Returns:
        str: Path to the output folder containing CLAHE-processed images.
    """

    # Set the output folder
    if output_folder is None:
        output_folder = os.path.join(images_folder, "processed")
    os.makedirs(output_folder, exist_ok=True)

    # List all JPEG image files in the input directory
    image_files = [f for f in os.listdir(images_folder)
                   if f.lower().endswith('.jpeg') and os.path.isfile(os.path.join(images_folder, f))]

    if not image_files:
        print(f"No supported image files found in {images_folder}.")
        return output_folder

    def apply_clahe(image_path, save_path):
        """
        Apply CLAHE to an image and save the result to disk.

        Parameters:
            image_path (str): Path to the input image.
            save_path (str): Path to save the processed image.
        """
        # Check if the processed image already exists
        if os.path.isfile(save_path):
            return  # Skip processing to save time

        # Read the image
        image = cv2.imread(image_path)
        if image is None:
            print(f"Warning: Failed to read {image_path}, skipping.")
            return

        try:
            # Convert to LAB color space
            lab = cv2.cvtColor(image, cv2.COLOR_BGR2LAB)
            l, a, b = cv2.split(lab)

            # Apply CLAHE to the L-channel
            clahe = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8, 8))
            l_clahe = clahe.apply(l)

            # Merge the CLAHE enhanced L-channel with A and B channels
            lab_clahe = cv2.merge((l_clahe, a, b))

            # Convert back to BGR color space
            final_image = cv2.cvtColor(lab_clahe, cv2.COLOR_LAB2BGR)

            # Save the processed image
            cv2.imwrite(save_path, final_image)
        except Exception as e:
            print(f"Error processing {image_path}: {e}")

    # Prepare list of tasks
    tasks = []
    for image_file in image_files:
        input_path = os.path.join(images_folder, image_file)
        output_path = os.path.join(output_folder, image_file)
        tasks.append((input_path, output_path))

    # Define a wrapper function for threading
    def process_task(task):
        input_path, output_path = task
        apply_clahe(input_path, output_path)

    # Use ThreadPoolExecutor for multithreading
    with ThreadPoolExecutor(max_workers=max_workers) as executor:
        # Submit all tasks to the executor
        futures = [executor.submit(process_task, task) for task in tasks]

        # Use tqdm to display progress bar
        for _ in tqdm(as_completed(futures), total=len(futures), desc="Processing Images with CLAHE", unit='images'):
            pass  # Progress is handled by tqdm

    print(f"All CLAHE-processed images are saved in: {output_folder}")

    # Display Examples
    try:
        # Validate DataFrame columns
        required_columns = {'image', 'level'}
        if not required_columns.issubset(df.columns):
            raise ValueError(f"DataFrame must contain columns: {required_columns}")

        # Select unique levels and sample images
        unique_levels = sorted(df["level"].unique())
        sampled_images = []

        for level in unique_levels:
            level_df = df[df["level"] == level]
            if not level_df.empty:
                image_name = level_df.iloc[0]["image"]
                sampled_images.append((image_name, level))

        if not sampled_images:
            print("No images found to display.")
            return output_folder

        # Calculate actual number of pairs to display
        actual_num_pairs = len(sampled_images)

        # Set up the matplotlib figure with num_pairs rows and 2 columns
        fig, axes = plt.subplots(actual_num_pairs, 2, figsize=(8, 3 * actual_num_pairs))

        # Handle the case when there's only one pair
        if actual_num_pairs == 1:
            axes = np.expand_dims(axes, axis=0)  # Convert to 2D array for consistency

        for idx, (image_name, level) in enumerate(sampled_images):
            # Define file paths (assuming JPEG format)
            original_path = os.path.join(images_folder, f"{image_name}.jpeg")
            processed_path = os.path.join(output_folder, f"{image_name}.jpeg")

            # Check if files exist
            if not os.path.isfile(original_path):
                print(f"Original image not found: {original_path}, skipping.")
                continue
            if not os.path.isfile(processed_path):
                print(f"Processed image not found: {processed_path}, skipping.")
                continue

            # Read images
            original = cv2.imread(original_path)
            processed = cv2.imread(processed_path)

            if original is None:
                print(f"Failed to read original image: {original_path}, skipping.")
                continue
            if processed is None:
                print(f"Failed to read processed image: {processed_path}, skipping.")
                continue

            # Convert BGR to RGB for matplotlib
            original_rgb = cv2.cvtColor(original, cv2.COLOR_BGR2RGB)
            processed_rgb = cv2.cvtColor(processed, cv2.COLOR_BGR2RGB)

            # Access the correct AxesSubplot objects
            ax_orig = axes[idx, 0]
            ax_proc = axes[idx, 1]

            # Plot Original Image
            ax_orig.imshow(original_rgb)
            ax_orig.set_title(f"Original - Level {level}", fontsize=14)
            ax_orig.axis("off")

            # Plot Processed Image
            ax_proc.imshow(processed_rgb)
            ax_proc.set_title(f"CLAHE Enhanced - Level {level}", fontsize=14)
            ax_proc.axis("off")

        # Adjust layout and display
        plt.tight_layout()
        plt.show()

    except Exception as e:
        print(f"An error occurred while displaying examples: {e}")

    return output_folder

processed_images_folder = process_and_display_images(preprocessed_folder, df)
```

```python
def calculate_exposure_metrics(processed_images_folder, df, over_threshold=200, under_threshold=50, well_exposed_threshold=60, max_workers=os.cpu_count() if os.cpu_count() else 4):
    """
    Calculates overexposure and underexposure metrics for images and adds them to the DataFrame in place.
    Defines the "Well Exposed" area using threshold lines instead of a rectangle.

    Parameters:
        processed_images_folder (str): Path to the folder containing CLAHE-processed images.
        df (pd.DataFrame): DataFrame containing at least an 'image' column with image names without extensions.
        over_threshold (int, optional): Pixel intensity threshold for overexposure. Defaults to 200.
        under_threshold (int, optional): Pixel intensity threshold for underexposure. Defaults to 50.
        well_exposed_threshold (int, optional): Threshold percentage defining "Well Exposed" area. Defaults to 60.
        max_workers (int, optional): Number of threads for parallel processing. Defaults to CPU count or 4.

    Returns:
        None: The DataFrame `df` is updated in place with new columns.
    """

    def process_image(row, images_folder, over_thresh, under_thresh):
        """
        Processes a single image to compute overexposure and underexposure metrics.

        Parameters:
            row (pd.Series): A row from the DataFrame containing 'image' and 'level'.
            images_folder (str): Path to the folder containing processed images.
            over_thresh (int): Pixel intensity threshold for overexposure.
            under_thresh (int): Pixel intensity threshold for underexposure.

        Returns:
            dict: Dictionary with 'image', 'over_threshold', 'under_threshold'.
        """
        image_name = row["image"]
        image_path = os.path.join(images_folder, f"{image_name}.jpeg")

        result = {"image": image_name,
                  "over_threshold": np.nan,
                  "under_threshold": np.nan}

        if not os.path.isfile(image_path):
            print(f"Warning: Image not found: {image_path}")
            return result  # Image not found

        try:
            # Load the image in grayscale
            gray_image = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)

            if gray_image is None:
                print(f"Warning: Failed to read image: {image_path}")
                return result  # Failed to read image

            # Total number of pixels
            total_pixels = gray_image.size

            # Number of pixels above over_threshold
            over_count = np.sum(gray_image >= over_thresh)

            # Number of pixels below under_threshold
            under_count = np.sum(gray_image <= under_thresh)

            # Calculate percentages
            over_percentage = (over_count / total_pixels) * 100
            under_percentage = (under_count / total_pixels) * 100

            # Assign results
            result["over_threshold"] = over_percentage
            result["under_threshold"] = under_percentage

            return result

        except Exception as e:
            print(f"Error processing {image_name}: {e}")
            return result  # Return NaNs on error

    # Initialize lists to collect results
    results = []

    # Initialize ThreadPoolExecutor for parallel processing
    with ThreadPoolExecutor(max_workers=max_workers) as executor:
        # Submit all image processing tasks to the thread pool
        futures = [executor.submit(process_image, row, processed_images_folder, over_threshold, under_threshold)
                   for _, row in df.iterrows()]

        # Use tqdm to display progress
        for future in tqdm(as_completed(futures), total=len(futures), desc="Calculating Exposure Metrics", unit="image"):
            res = future.result()
            results.append(res)

    # Convert results to DataFrame
    results_df = pd.DataFrame(results)

    if results_df.empty:
        print("No exposure metrics were computed.")
        return

    # Create a mapping from 'image' to 'over_threshold' and 'under_threshold'
    over_mapping = results_df.set_index('image')['over_threshold'].to_dict()
    under_mapping = results_df.set_index('image')['under_threshold'].to_dict()

    # Add new columns to the original df in place
    df['over_threshold'] = df['image'].map(over_mapping)
    df['under_threshold'] = df['image'].map(under_mapping)

    print("Added 'over_threshold' and 'under_threshold' columns to the DataFrame.")

    # --- Plotting the Exposure Metrics ---
    # Check if there are any valid exposure metrics to plot
    if results_df[['over_threshold', 'under_threshold']].notnull().any().any():
        plt.figure(figsize=(12, 6))

        # Determine if 'level' exists for hue differentiation
        if 'level' in df.columns:
            scatter = sns.scatterplot(
                data=df,
                x='under_threshold',
                y='over_threshold',
                hue='level',
                palette='coolwarm',
                alpha=0.7,
                edgecolor='k')
        else:
            scatter = sns.scatterplot(
                data=df,
                x='under_threshold',
                y='over_threshold',
                color='blue',
                alpha=0.7,
                edgecolor='k')

        # Define the "Well Exposed" threshold
        threshold = well_exposed_threshold  # e.g., 60%

        # Add horizontal and vertical threshold lines without adding to the legend
        plt.axhline(y=threshold, color='green', linestyle='--', linewidth=2, label='_nolegend_')
        plt.axvline(x=threshold, color='green', linestyle='--', linewidth=2, label='_nolegend_')

        # Annotate the "Well Exposed" area at the lower-left of the intersection
        plt.text(threshold - 5, threshold - 5, 'Well Exposed', color='green', fontsize=12,
                 horizontalalignment='right', verticalalignment='top',
                 bbox=dict(facecolor='white', alpha=0.6, edgecolor='none'))

        # Set plot titles and labels with increased font sizes
        plt.title("Overexposure vs Underexposure Metrics with Well Exposed Threshold", fontsize=16)
        plt.xlabel("Underexposure Percentage (%)", fontsize=14)
        plt.ylabel("Overexposure Percentage (%)", fontsize=14)

        # Customize legend
        if 'level' in df.columns:
            # Get existing handles and labels
            handles, labels = scatter.get_legend_handles_labels()
            plt.legend(title="Level", bbox_to_anchor=(1.05, 1), loc='upper left', fontsize=12)
        else:
            # So no legend entry is needed
            plt.legend().remove()

        # Add grid for better readability
        plt.grid(True, alpha=0.7)

        # Set limits to ensure zero is not exactly on the plot edge and axes range from 0 to 100
        plt.xlim(-5, 105)
        plt.ylim(-5, 105)

        # Set ticks every 10 units
        plt.xticks(ticks=np.arange(0, 101, 10))
        plt.yticks(ticks=np.arange(0, 101, 10))

        # Improve layout to prevent clipping of labels/legends
        plt.tight_layout()
        # Display the plot
        plt.show()
    else:
        print("No valid exposure metrics available for plotting.")

calculate_exposure_metrics(images_path+"/preprocessed/processed", df, over_threshold = 190, under_threshold = 25)
```

```python
# Config
IMAGES_FOLDER = images_path + "/preprocessed/processed"
RADIUS = 3
N_POINTS = 8 * RADIUS
HOG_PIXELS_PER_CELL = (16, 16)
HOG_CELLS_PER_BLOCK = (2, 2)
HOG_ORIENTATIONS = 9

def extract_features(image_path):
    img = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)
    if img is None:
        return None

    # LBP
    lbp = local_binary_pattern(img, N_POINTS, RADIUS, method='uniform')
    lbp_hist, _ = np.histogram(
        lbp.ravel(),
        bins=np.arange(0, N_POINTS+3),
        range=(0, N_POINTS+2)
    )
    lbp_hist = lbp_hist.astype("float")
    lbp_hist /= (lbp_hist.sum() + 1e-7)

    # HOG
    hog_features = hog(
        img,
        orientations=HOG_ORIENTATIONS,
        pixels_per_cell=HOG_PIXELS_PER_CELL,
        cells_per_block=HOG_CELLS_PER_BLOCK,
        block_norm='L2-Hys',
        transform_sqrt=True,
        feature_vector=True
    )

    return np.hstack([lbp_hist, hog_features])

def parallel_feature_extraction(row, folder):
    index, row_data = row
    img_path = os.path.join(folder, f"{row_data['image']}.jpeg")
    feats = extract_features(img_path)
    return feats, row_data['level']

def build_features_in_parallel(df, folder, max_workers=os.cpu_count()):
    feats_list, labels_list = [], []
    with ThreadPoolExecutor(max_workers=max_workers) as executor:
        results = list(tqdm(
            executor.map(lambda r: parallel_feature_extraction(r, folder), df.iterrows()),
            total=len(df),
            desc="Parallel Feature Extraction"
        ))
    for feats, label in results:
        if feats is not None:
            feats_list.append(feats)
            labels_list.append(label)
    return np.array(feats_list), np.array(labels_list)


# 1) Build features
X, y = build_features_in_parallel(df, IMAGES_FOLDER)

# 2) Split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=random_state)
```
